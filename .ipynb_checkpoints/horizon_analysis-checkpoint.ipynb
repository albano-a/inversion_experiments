```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
cubic = "./saida/horizon_cubic.npy"
idw = "./saida/horizon_idw.npy"
kriging = "./saida/horizon_kriging.npy"
linear = "./saida/horizon_linear.npy"
nearest = "./saida/horizon_nearest.npy"
rbf = "./saida/horizon_rbf.npy"
```

```python
cubic = np.load(cubic)
idw = np.load(idw)
kriging = np.load(kriging)
linear = np.load(linear)
nearest = np.load(nearest)
rbf = np.load(rbf)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(10, 7))

axes[0, 0].imshow(cubic)
axes[0, 1].imshow(idw)
axes[0, 2].imshow(kriging)
axes[1, 0].imshow(linear)
axes[1, 1].imshow(nearest)
axes[1, 2].imshow(rbf)

plt.show()
```

```python
def plot_horizons(surface):
    plt.imshow(surface, cmap="gist_rainbow_r")
    plt.colorbar()
    plt.show()
```

```python
surfaces = [cubic, idw, linear, kriging, rbf, nearest]

for s in surfaces:
    plot_horizons(s)
```
